```python
#task 1
inp1="BaNaNa"
countU=0
countL=0
for i in inp1:
    if 65<=ord(i)<=90:
        countU+=1
    else:
        countL+=1
if countU>countL:
    print(inp1.upper())
else:
    print(inp1.lower())
```

```python
#task 2
inp1="jhg231j213"
countN=0
countW=0
for i in inp1:
    if 48<=ord(i)<=57:
        countN+=1
    else:
        countW+=1
if countN!=0 and countW==0:
    print("NUMBER")
elif countN==0 and countW!=0:
    print("WORD")
else:
    print("MIXED")
```

```python
#task 5
inp1="ohMyBR@CU"
countU=0
countL=0
countD=0
countS=0
str1=""
for i in inp1:
    if "A"<=i<="Z":
        countU+=1
    elif "a"<=i<="z":
        countL+=1
    elif "0"<=i<="9":
        countD+=1
    else:
        countS+=1
if countU==0:
    str1+="Uppercase character missing, "
if countL==0:
    str1+="Lowercase character missing, "
if countD==0:
    str1+="Digit missing, "
if countS==0:
    str1+="Special character missing, "
if countU!=0 and countL!=0 and countD!=0 and countS!=0:
    str1+="OK  "
print(str1[:-2:])
```

```python
#task 1
list1=[]
while True:
    x=input()
    if x=="STOP":
        break
    else:
        list1.append(int(x))
list2=[]
for i in list1:
    if i not in list2:
        list2.append(i)
for k in list2:
    print(f"{k} - {list1.count(k)} times")
```

```python
#Task 4
while True:
    a = input()
    if a == "STOP":
        break
    list1 = [int(b) for b in a.split()]
    list2 = []
    diff=0
    for i in range(1, len(list1)):
        diff = (list1[i] - list1[i - 1])
        if diff<0:
          diff=diff*(-1)
          list2.append(diff)
        else:
           list2.append(diff)
    for i in range(1, len(list1) - 1):
        if i not in list2:
            flag = False
            break
        else:
            flag = True
    if flag==True:
        print("UB Jumper")
    else:
        print("Not UB Jumper")
```

```python
#Task 1
dict1={key:int(value) for key, value in (pair.split(': ') for pair in input().split(', '))}
dict2={key:int(value) for key, value in (pair.split(': ') for pair in input().split(', '))}
dict3={}
list1=[]
for k,v in dict1.items():
    if k in dict2:
        dict3[k] = dict1[k] + dict2[k]
    else:
        dict3[k] = v
for k,v in dict2.items():
    if k not in dict3:
        dict3[k] = v
for v in dict3.values():
    if v not in list1:
        list1.append(v)
list1.sort()
print(dict3)
print(f"Values: {tuple(list1)}")
```

```python
#Task 3
inp1=input()   # key1 : value1, key2 : value2, key3 : value1
list1=inp1.split(", ")
#print(list1)
dict1={}
for i in list1:
    list2=i.split(" : ")
    key=list2[0]
    val=list2[-1]
    dict1[key]=val
#print(dict1)
dict2={}
for k,v in dict1.items():
    if v not in dict2.keys():
        dict2[v]=[k]
    else:
        dict2[v].append(k)
print(dict2)
```
