```python
#Task 1
def bmi(tup1):
    height,weight=tup1
    h=height*0.01
    w=weight
    b=round(w/(h**2),1)    #round(calculation, n decimal)
    if b<18.5:
        print(f"Score is {b}. You are Underweight")
    elif 18.5<=b<=24.9:
        print(f"Score is {b}. You are Normal")
    elif 25<=b<=30:
        print(f"Score is {b}. You are Overweight")
    else:
        print(f"Score is {b}. You are Obese")
bmi((175,96))
bmi((152,48))
```

```python
#Task 2
def order(burger,loc="Mohakhali"):
    meal=0
    delivery=0
    tax=0
    if burger=="BBQ Chicken Cheese Burger":
        meal=250
    elif burger=="Beef Burger":
        meal=170
    else:
        meal=200
    if loc=="Mohakhali":
        delivery=40
    else:
        delivery=60
    tax=meal*0.08
    total=meal+delivery+tax
    print(total)
order("Beef Burger","Dhanmondi")
order("Beef Burger")
```

```python
#Task 3
def replace_domain(x,y,z="sheba.xyz"):
    counter=0
    str1=""
    str2=y
    str3=""
    if z=="sheba.xyz":
        print(f"unchanged: {x}")
    if z=="kaaj.com":
        for i in x:
            counter+=1
            if i=="@":
                break
        for j in range(counter):
            str1+=x[j]
        str3=str1+str2
        print(f"Changed: {str3}")
replace_domain("alice@kaaj.com","sheba.xyz","kaaj.com")
replace_domain("bob@sheba.xyz","sheba.xyz")
```

```python
#Task 4
def palindrome(str1):
    str2=""
    for i in str1:
        if i==" ":
            continue
        else:
            str2+=i
    str3=str2[::-1]
    if str2==str3:
        print("Palindrome")
    else:
        print("Not a palindrome")
palindrome("madam")
palindrome("hello")
palindrome("nurses run")
```

```python
#Task 5
def days(num):
    years=num//365
    months=(num-(years*365))//30
    days=(num-(years*365)-(months*30))
    print(f"{years} years, {months} months and {days} days")
days(4320)
days(4000)
```
