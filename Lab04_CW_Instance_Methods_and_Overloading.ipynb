```python
#Task 01
class Customer:
    def __init__(self):
        print(f"Welcome to ABC Memorial Park")
        self.count=0
        self.total=0
    def buyTicket(self,name,age):
        self.count+=1                   # Important
        if self.count>3:
            print(f"You can't buy more than 3 tickets")
            self.count=3                # Important
        else:
            if age>10:
                print(f"Successfully purchased a ticket for {name}!")   # {name} Important
                self.total+=100
            else:
                print(f"Successfully purchased a ticket for {name}!")
                self.total+=50
    def showDetails(self):
        print(f"Amount of tickets: {self.count} \nTotal price: {self.total} Taka")
print('1-------------------------')
customer1 = Customer()
print('2-------------------------')
customer1.buyTicket('Bob', 23)
customer1.buyTicket('Henry', 7)
customer1.buyTicket('Alexa', 30)
customer1.buyTicket('Jonas', 43)
print('3-------------------------')
customer1.showDetails()
print('4-------------------------')
customer2 = Customer()
print('5-------------------------')
customer2.buyTicket('Harry', 60)
customer2.buyTicket('Tomas', 28)
print('6-------------------------')
customer2.showDetails()
```

```python
#Task 2
class SentenceAnalyzer:
    def __init__(self,str1=""):
        self.str1=str1
        self.count=0
    def set_sentence(self,str2):
        if self.str1=="":
            self.str1=str2
    def get_word_count(self,*args):
        list1=[]
        if args==():
            list1=[i for i in self.str1.split(" ")]
            print(f"Number of words in the sentence: {len(list1)}")
        else:
            count=0
            list1=[i for i in self.str1.split(" ")]
            for j in list1:
                if len(j)==args[-1]:        # Important   (4,)   kintu -1 dile 4 ashbe
                    count+=1
            print(f"Number of words with {args[-1]} characters in the sentence: {count}")
analyzer1 = SentenceAnalyzer()
analyzer1.set_sentence("That's an easy one")
print("1--------------------------")
analyzer1.get_word_count()
print("2--------------------------")
analyzer2 = SentenceAnalyzer("Like I said it's easy")
print("3--------------------------")
analyzer2.get_word_count()
print("4--------------------------")
analyzer2.get_word_count(4)
print("5--------------------------")
analyzer1.get_word_count(5)
```

```python
#Task 03
class Student:
  def __init__(self,name,cg,credit=9,dept="CSE"):
    self.nam=name
    self.cg=cg
    self.credit=credit
    self.dept=dept
    self.sch=""
    self.sch2=""
  def showDetails(self):
    print(f"Name:{self.nam}")
    print(f"Department:{self.dept}")
    print(f"CGPA:{self.cg}")
    print(f"Number of Credits: {self.credit}")
    print(f"Scholarship Status: {self.sch2}")
  def checkScholarshipEligibility(self):
    if self.cg>3.7 and self.credit>10:
      self.sch="eligible for Merit-based scholarship"
      self.sch2="Merit-based scholarship"
    elif self.cg>=3.5 and self.cg<=3.7 and self.credit>10:
      self.sch="eligible for Need-based scholarship."
      self.sch2="Need-based scholarship "
    elif self.cg>=3.5 and self.credit>10:
      self.sch="eligible for scholarship"
    else:
      self.sch="Not eligible for scholarship"
      self.sch2="No scholarship"
    print(self.nam,"is",self.sch)

print('--------------------------')
std1 = Student("Alif",3.99,12)
print('--------------------------')
std1.checkScholarshipEligibility()
print('--------------------------')
std1.showDetails()
print('--------------------------')
std2 = Student("Mim",3.4)
std3 = Student("Henry",3.5,15,"BBA")
print('--------------------------')
std2.checkScholarshipEligibility()
print('--------------------------')
std3.checkScholarshipEligibility()
print('--------------------------')
std2.showDetails()
print('--------------------------')
std3.showDetails()
print('--------------------------')
std4 = Student("Bob",4,6,"CSE")
print('--------------------------')
std4.checkScholarshipEligibility()
print('--------------------------')
std4.showDetails()
```

```python
#Task 05
class TaxiLagbe:
    def __init__(self, num, area):
        self.num = num
        self.area = area
        self.ride = []
        self.price = 0
    def addPassenger(self, *tup1):
        for i in tup1:
            x = i.split('_')
            if len(self.ride)<4:
                self.ride.append(x[0])
                print(f'Dear {x[0]}! Welcome to TaxiLagbe')
                self.price += int(x[1])
            else:
                print('Taxi Full! No more passengers can be added.')
    def printDetails(self):
        print(f'Trip info for Taxi number: {self.num}')
        print(f'This taxi can only cover the {self.area} area.')
        print(f'Total passenger: {len(self.ride)}')
        print(f'Passenger list:')
        a = ''
        for i in self.ride:
            a += i + ', '
        print(a[:-2:])
        print(f'Total collected fare: {self.price} Taka')


# Do not change the following lines of code.
taxi1 = TaxiLagbe('1010-01', 'Dhaka')
print('-------------------------------')
taxi1.addPassenger('Walker_100', 'Wood_200','Matt_100')
taxi1.addPassenger('Wilson_105')
print('-------------------------------')
taxi1.printDetails()
print('-------------------------------')
taxi1.addPassenger('Karen_200')
print('-------------------------------')
taxi1.printDetails()
print('-------------------------------')
taxi2 = TaxiLagbe('1010-02', 'Khulna')
taxi2.addPassenger('Ronald_115', 'Parker_215')
print('-------------------------------')
taxi2.printDetails()
```
