```python
# Task 02
class Foodie:
    def __init__(self, name):
        self.name = name
        self.items = []
        self.total_spent = 0

    def order(self, *items):
        for i in items:
            total = 0
            list1 = i.split('-')
            if list1[0] in menu:
                self.items.append(list1[0])
                total += (menu[list1[0]]) * int(list1[1])
            self.total_spent += total
            print(f'Ordered - {list1[0]}, quantity - {list1[1]}, price (per Unit) - {menu[list1[0]]}.\n'
                f'Total price - {total}')

    def show_orders(self):
        output = f'{self.name} has {len(self.items)} item(s) in the cart.\n'f'Items: {self.items}\n'f'Total spent: {self.total_spent}.'
        return output

    def pay_tips(self, amount = 0):
        self.total_spent += amount
        if amount == 0:
            print('No tips to the waiter.')
        else:
            print(f'Gives {amount}/- tips to the waiter.')



menu = {'Chicken Lollipop':15,'Beef Nugget':20,'Americano':180,'Red Velvet':150,'Prawn Tempura':80,'Saute Veg':200}

f1 = Foodie('Frodo')
print(f1.show_orders())
print('1----------------------')
f1.order('Chicken Lollipop-3','Beef Nugget-6','Americano-1')
print('2----------------------')
print(f1.show_orders())
print('3----------------------')
f1.order('Red Velvet-1')
print('4----------------------')
f1.pay_tips(20)
print('5----------------------')
print(f1.show_orders())
f2 = Foodie('Bilbo')
print('6----------------------')
f2.order('Prawn Tempura-6','Saute Veg-1')
print('7----------------------')
f2.pay_tips()
print('8----------------------')
print(f2.show_orders())
```

```python
# Task 03
class Farmer:
    def __init__(self, name=None):
        self.name = name
        self.crops = []
        self.fishes = []

        if name == None:
            print(f'Welcome to your farm!')
        elif isinstance(name, int):
            print(f"Welcome to your farm. Your farm ID is {self.name}!")
        else:
            print(f'Welcome to your farm, {self.name}')

    def addCrops(self, *crops):
        self.crops.extend(crops)
        if self.crops:
            print(f'{len(crops)} crop(s) added.')
        else:
            print("No crop(s) added.")

    def addFishes(self, *fishes):
        self.fishes.extend(fishes)
        if self.fishes:
            print(f'{len(fishes)} fish(s) added.')
        else:
            print("No fish(s) added.")

    def showGoods(self):
        if self.crops:
            print(f"You have {len(self.crops)} crop(s).")
            print(','.join(self.crops))
        else:
            print("You don't have any crop(s).")
        if self.fishes:
            print(f"You have {len(self.fishes)} fish(s).")
            print(','.join(self.fishes))
        else:
            print("You don't have any fish(s).")



f1 = Farmer()
print("-------------------")
f1.addCrops('Rice', "Jute", "Cinnamon")
print("-------------------")
f1.addFishes()
print("-------------------")
f1.addCrops('Mustard')
print("-------------------")
f1.showGoods()
print("-------------------")
f2 = Farmer("Korim Mia")
print("-------------------")
f2.addFishes('Pangash', 'Magur')
print("-------------------")
f2.addCrops("Wheat", "Potato")
print("-------------------")
f2.addFishes("Koi", "Tuna", "Sardine")
print("-------------------")
f2.showGoods()
print("-------------------")
f3 = Farmer(2865127000)
print("-------------------")
f3.addCrops()
print("-------------------")
f3.addFishes("Katla")
print("-------------------")
f3.showGoods()
print("-------------------")
```

```python
# Task 04
class UniversalStudiosUser:
    def __init__(self,name,age:int,user_type):
        self.name = name
        self.age = age
        self.user_type = user_type
        print("Welcome to Universal Studios.")

    def selected_rides(self,*rides):
        self.rides = rides
        print("Added rides successfully.")

    def show_details(self):
        subtotal = 0
        discount = 0
        total_cost = 0
        print(f'Your information: ')
        print(f'Name: {self.name}, Age: {self.age}, Category: {self.user_type}\nSelected rides:')

        if self.user_type == "Normal":
            for ride in self.rides:
                ride_name, cost = ride.split("-")
                print(f"Ride: {ride_name}, Amount: {cost} dollar(s)")
                total_cost += int(cost)

        elif self.user_type == "Special" and len(self.rides) > 3:
            for ride in self.rides:
                ride_name, cost = ride.split("-")
                print(f"Ride: {ride_name}, Amount: {cost} dollar(s)")
                subtotal += int(cost)
            discount += (20/100)*subtotal
            total_cost += subtotal - discount
            print(f"Congratulations!!! You've got a {int(discount/11)}% discount.")

        elif self.user_type == "Special":
            for ride in self.rides:
                ride_name, cost = ride.split("-")
                print(f"Ride: {ride_name}, Amount: {cost} dollar(s)")
                total_cost += int(cost)


        print(f'Please pay {total_cost} dollar(s).')

customer_1 = UniversalStudiosUser("Alice", 21, "Special")
print("--------- 1 ---------")
customer_1.selected_rides("Waterworld-100", "Accelerator-200", "DinoSoarin-50")
print("--------- 2 ---------")
customer_1.show_details()


print("=================")


customer_2 = UniversalStudiosUser("Bob", 20, "Normal")
print("--------- 3 ---------")
customer_2.selected_rides("Enchanted Airways-300", "Jurassic Park-500", "Accelerator-200", "DinoSoarin-50")
print("--------- 4 ---------")
customer_2.show_details()


print("=================")


customer_3 = UniversalStudiosUser("Mark", 15, "Special")
print("--------- 5 ---------")
customer_3.selected_rides("Transformers-450", "Jurassic Park-500", "Waterworld-100", "DinoSoarin-50")
print("--------- 6 ---------")
customer_3.show_details()
```

```python
# Task 05
class Department:
    def __init__(self,dept="ChE Department",sec_num=5):
        self.dept = dept
        self.sec_num = sec_num
        self.average_students = 0
        print(f"The {self.dept} has {self.sec_num} sections.")

    def add_students(self,*student_num):
        self.student_num = student_num
        if len(student_num) == self.sec_num:
            self.average_students += sum(student_num) / self.sec_num
            print(f"The {self.dept} has an average of {self.average_students} students in each section.")
            self.Total_student = self.average_students * self.sec_num

        else:
            print(f"The {self.dept} doesn't have {len(self.student_num)} sections.")
    def merge_Department(self, *dept_obj):
        for d in dept_obj:
            print(f'{d.dept} is merged to {self.dept}.')
            self.Total_student += d.average_students * d.sec_num
        avg_mega = self.Total_student / self.sec_num
        return f'Now the {self.dept} has an average of {avg_mega} students in each section.'


d1 = Department()
print('1-----------------------------------')
d2 = Department('MME Department')
print('2-----------------------------------')
d3 = Department('NCE Department', 8)
print('3-----------------------------------')
d1.add_students(12, 23, 12, 34, 21)
print('4-----------------------------------')
d2.add_students(40, 30, 21)
print('5-----------------------------------')
d3.add_students(12, 34, 41, 17, 30, 22, 32, 51)
print('6-----------------------------------')
mega = Department('Engineering Department', 10)
print('7-----------------------------------')
mega.add_students(21,30,40,36,10,32,27,51,45,15)
print('8-----------------------------------')
print(mega.merge_Department(d1, d2))
print('9-----------------------------------')
print(mega.merge_Department(d3))
```

```python
# Task 6
class StudentRoutineGenerator:
    def __init__(self, name, max_courses):
        self.name = name
        self.max_courses = max_courses
        self.routine = {
            'Sat/Thurs': {},
            'Sun/Tue': {},
            'Mon/Wed': {}
        }
        self.count = 0
        self.courses = []
        print(f'Name: {self.name}\n'
              f"Maximum number of Courses: {self.max_courses}\n"
              f"Initial routine: {self.routine}")

    def addCourses(self, *routine):
        for elem in routine:
            list1 = elem.split('-')
            course = list1[0]
            if course not in self.courses:
                self.courses.append(course)
            else:
                print(f"You already have {course} in your routine.")
                continue
            day = list1[1]
            time = list1[2]

            if day in self.routine:
                if time not in self.routine[day]:
                    if self.count < self.max_courses:
                        self.count += 1
                        self.routine[day][time] = course
                        print(f"Sucessfully added {course}!")
                    else:
                        print(f"You can't take more than {self.max_courses} courses.")
                else:
                    print(f"Can't take {course}. It's clashing with your {self.routine[day][time]}")

    def dropCourse(self, course):
        flag = False
        for value in self.routine.values():
            for t, c in value.items():
                if course == c:
                    temp = t
                    flag = True
                    break
                else:
                    flag = False
        if flag == True:
            value.pop(temp)
            print(f"Successfully dropped {course}.")
        else:
            print("No such course in your routine.")

    def showRoutine(self):
        print(f"Updated Routine: {self.routine}")
        print("Routine Details:")
        for key, value in self.routine.items():
            if value != {}:
                print(f"{key}:")
            for time, course in value.items():
                print(f"{time} - {course}")



print('##################################')
st1 = StudentRoutineGenerator('Harry', 4)
print('------------------------------')
st1.addCourses('CSE110-Mon/Wed-12:30', 'MAT110-Mon/Wed-2:00')
st1.addCourses('ENG101-Sun/Tue-12:30', 'CSE110-Mon/Wed-9:30')
st1.addCourses('PHY111-Sun/Tue-12:30')
print('------------------------------')
st1.showRoutine()
print('------------------------------')
st1.dropCourse('CSE110')
st1.dropCourse('PHY112')
print('------------------------------')
st1.showRoutine()
print('##################################')
st2 = StudentRoutineGenerator('John', 3)
print('------------------------------')
st2.addCourses('MAT110-Mon/Wed-8:00')
st2.addCourses('ENG101-Sat/Thurs-12:30', 'CSE110-Sun/Tue-9:30')
st2.addCourses('PHY111-Sun/Tue-12:30')
print('------------------------------')
st2.showRoutine()
```
