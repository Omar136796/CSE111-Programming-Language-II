```python
#Task 02
class Spaceship:
    def __init__(self,n1,cap):
        self.n1=n1
        self.cap=cap
        self.total=0
        self.listC=[]
    def load_cargo(self,n3):
        sum=self.total+n3.getWeight()
        if sum<=self.cap:
            self.listC.append(n3.getName())
            self.total+=n3.getWeight()
        else:
            print(f"Warning: Unable to load {n3.getName()} inside ")
            print(f"{self.n1}. Exceeds capacity by {sum-self.cap}.")
    def display_details(self):
        print(f"Spaceship Name: {self.n1}")
        print(f"Capacity: {self.cap}")
        print(f"Current Cargo Weight: {self.total}")
        print(f"Cargo: {self.listC}")
class Cargo:
    def __init__(self,n2,weight):
        self.__n2=n2
        self.__weight=weight
    def getName(self):
        return self.__n2
    def getWeight(self):
        return self.__weight
# Creating spaceships
falcon = Spaceship("Falcon", 50000)
apollo = Spaceship("Apollo", 100000)
enterprise = Spaceship("Enterprise", 220000)
print("1.===================================")
# Creating cargo
gold = Cargo("Gold", 20000)
platinum = Cargo("Platinum", 25000)
dilithium = Cargo("Dilithium", 50000)
trilithium = Cargo("Trilithium", 70000)
neutronium = Cargo("Neutronium", 80000)
print("2.===================================")
# Loading cargo onto spaceships
falcon.load_cargo(gold)
falcon.load_cargo(platinum)
falcon.display_details()
print("3.===================================")
apollo.load_cargo(gold)  # Apollo will not reach its total capacity
apollo.display_details()
print("4.===================================")
falcon.load_cargo(neutronium)  # This should exceed Falcon's capacity
print("5.===================================")
enterprise.load_cargo(dilithium)
enterprise.load_cargo(trilithium)
enterprise.load_cargo(neutronium)  # This should not exceed Enterprise's capacity
enterprise.display_details()
```

```python
# Task 03

class Song:

    def __init__(self, title , duration , artist = 'Unknown Artist'):
        self.__title = title
        self.__duration = duration
        self.__artist = artist

    def get_title(self):
        return self.__title
    def get_artist(self):
        return self.__artist
    def get_duration(self):
        return self.__duration


    def set_title(self,a):
        self.__title =a
    def set_artist(self,a):
        self.__artist =a
    def set_duration(self,a):
        self.__duration =a


    def info(self):
        if self.__artist != 'Unknown Artist':
            return f"{self.get_title()} by {self.get_artist()} ({self.get_duration()// 60}: {self.get_duration() % 60})"
        elif self.__artist == 'Unknown Artist':
            return f"{self.__title} by {self.__artist} ({self.__duration // 60}: {self.__duration % 60})"


class Playlist:

    def __init__(self,name, song_list = [], now_id = -1):
        self.__name = name
        self.__song_list = []
        if song_list != []:
            for i in song_list:
                self.__song_list.append(i)
        self.__now_id = now_id
        self.nos = len(song_list)


    def get_name(self):
        return self.__name
    def get_song_list(self):
        return self.__song_list
    def get_now_id(self):
        return self.__now_id


    def set_name(self,a):
        self.__name =a
    def set_song_list(self,*a):
        self.__song_list =list(a)
    def set_now_id(self,a):
        self.__now_id =a

    def add_songs(self, *songs):
        c = 0
        for i in songs:
            self.nos += 1
            self.__song_list.append(i)
            c+= 1
        print(f"{c} songs added to the playlist")

    def remove_song(self,song_obj):
        self.nos -= 1
        self.__song_list.remove(song_obj)


    def play_next(self):
        if self.__now_id == -1:
            self.__now_id += 1
        else:
            if self.__now_id == len(self.__song_list) -1:
                print("reached the end of the playlist. Starting from the first song")
                self.__now_id =0
            else:
                self.__now_id +=1

        print(f"Now playing {self.__song_list[self.__now_id].info()}")

    def play_prev(self):
        if self.__now_id == 0:
            print(f"Cannot go back from the first song.")
            self.__now_id = 0

        else:
            self.__now_id -= 1

        print(f"Now playing {self.__song_list[self.__now_id].info()}")

    def show_playlist(self):
        print(f"Playlist: {self.get_name()}")
        print(f"Total Songs: {self.nos}")
        for i in range(len(self.__song_list)):
            print(f"{i+1}. {self.__song_list[i].info()}")

print("---------------------")
s1 = Song('The ghost of you',75,'My Chemical Romance')
print(s1.info())
print("---------------------")
s2 = Song('Prostab',95,'Odd signature')
print(s2.info())
print("---------------------")
s3 = Song('Amar ache jol',115,'Shawon')
print(s3.info())
print("---------------------")
s4 = Song('Never gonna give you up',135,'Rick Astley')
print(s4.info())
print("---------------------")
s5 = Song('Ei bidaye',155,'Artcell')
print(s5.info())
print("---------------------")

p1 = Playlist('My playlist', [s1,s2])
p1.show_playlist()

print("---------------------")
p1.add_songs(s3,s4)
p1.show_playlist()

print("---------------------")
p1.remove_song(s3)
p1.show_playlist()
print("---------------------")
print("---------------------")

p1.play_next()
p1.play_next()
p1.play_next()
p1.play_next()

print("---------------------")
print("---------------------")
p1.play_next()
p1.play_next()

print("---------------------")
print("---------------------")
p1.play_prev()
p1.play_prev()
p1.play_prev()
```
