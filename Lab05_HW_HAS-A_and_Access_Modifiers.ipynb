```python
class Book:

    def __init__(self, title, author, genre):
        self.__title = title
        self.__author = author
        self.__genre = genre
        self.__available = True
        self.__borrower = None


    def set_title(self, a):
        self.__title = a

    def get_title(self):
        return self.__title

    def set_author(self, a):
        self.__author = a

    def get_author(self):
        return self.__author

    def set_genre(self,a):
        self.__genre = a

    def get_genre(self):
        return self.__genre


    def set_availability(self, a):
        self.__available = a

    def get_availability(self):
        return self.__available

    def set_borrower(self,a):
        self.__borrower = a

    def get_borrower(self):
        return self.__borrower


    def display_info(self):
        x = ''
        x += f'Book Name: {self.__title}\n'
        x += f'Author Name: {self.__author}\n'
        x += f'Genre Name: {self.__genre}\n'
        x += f'Availability Status : {self.__available}\n'
        x += f"Borrowed by {self.__borrower}"

        return x


class LibraryMember:

    def __init__(self,ID, name):
        self.__id = ID
        self.__name = name
        self.borrowed_books = []

    def set_id(self, a):
        self.__id = a

    def get_id(self):
        return self.__id

    def set_name(self, a):
        self.__name = a

    def get_name(self):
        return self.__name

    def borrow_book(self, book):
        self.borrowed_books.append(book)
        book.set_availability(False)
        book.set_borrower(self.__id)


    def return_book(self, book):
        self.borrowed_books.remove(book)
        book.set_availability(True)
        book.set_borrower(None)

    def display_borrowed_books(self):
        print(f"Books borrowed by: {self.__name}")
        for i in self.borrowed_books:
            print(i.display_info())
            print('_______________')

class Library:

    def __init__(self):
        self.books_available = []
        self.library_members = []


    def add_book(self, book):
        self.books_available.append(book)

    def add_library_member(self, member):
        self.library_members.append(member)

    def display_book_list(self):
        print(f"All the books in library are: ")
        for i in self.books_available:
            print(i.display_info())
            print('__________________')

    def display_library_members(self):
        print(f"All the members in the library are: ")
        for i in self.library_members:
            print(i.get_name())
            print('_______________')



print('--------------------------------')
book1 = Book('Holud himu kalo rab','Humaiyun Ahmed','fiction')
book2 = Book('A brief story of time','Stephen Hawking','Science')
book3 = Book('Calculas I','Howard Anton','Education')
book4 = Book('Hazar bochor dhore','Jahir Rayhan','Social')
book5 = Book('Hangor nodi grenade','Selina Hossain','War')

print(book1.display_info())
print('--------------------------------')
print(book2.display_info())
print('--------------------------------')
print(book3.display_info())
print('--------------------------------')
print(book4.display_info())
print('--------------------------------')
print(book5.display_info())
print('#################################')
print('#################################')
print('#################################')


member1 = LibraryMember('id1','Faruk')
member2 = LibraryMember('id2','Omar')

member1.borrow_book(book1)
print(book1.display_info())
print('--------------------------------')
member1.return_book(book1)
print(book1.display_info())
print('--------------------------------')
print('--------------------------------')

member2.borrow_book(book3)
member2.borrow_book(book5)
member2.display_borrowed_books()
print('--------------------------------')
member2.return_book(book3)
print(book1.display_info())


print('#################################')
print('#################################')
print('#################################')

member1.borrow_book(book2)
member1.borrow_book(book4)
member1.display_borrowed_books()

print('#################################')
print('#################################')
print('#################################')

library1 = Library()
library1.add_book(book1)
library1.add_book(book2)
library1.add_book(book3)
library1.add_book(book4)
library1.add_book(book5)

library1.add_library_member(member1)
library1.add_library_member(member2)

library1.display_book_list()
print('--------------------------------')
library1.display_library_members()
```
