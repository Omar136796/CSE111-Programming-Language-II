```python
class Movie:
    def __init__(self, name, genre, duration):
        self.name = name
        self.genre = genre
        self.duration = duration

    def movieInfo(self):
        s = f"Movie Name: {self.name}\nMovie Genre: {self.genre}\nDuration: {self.duration}"
        return s
    @classmethod
    def createMovie_fromString(cls,s):
        x,y,z = s.split('-')
        z = int(z)
        t = cls(x,y,z)
        return t

class StarCinema:
    all_branch_info = {}

    def __init__(self, b_name):
        if b_name in StarCinema.all_branch_info:
            pass
        else:
            self.b_name = b_name
            self.m_list = []
            StarCinema.all_branch_info[self.b_name] = self.m_list
            print(f"Welcome to the {b_name} branch of StarCinema! ")

    def addMovies(self,*m):
        m = list(m)
        for i in m:
            if i in self.m_list:
                print(f"Movie is already added in this branch.")
            else:
                self.m_list.append(i)
                print(f"{i.name} added to {self.b_name} branch.")
    def removeMovie(self,m):
        if m in self.m_list:
            self.m_list.remove(m)
        else:
            print(f"Failed to remove movie {m}")
    @staticmethod
    def check(n):
        s = 0
        z = 0
        for k,v  in StarCinema.all_branch_info.items():
            for i in v:
                if n == i.name:
                    print(f"{n} is being streamed in {k} branch.")
                    s = 1
                    z = i
        if s==0:
            print(f"{n} not showing")
        else:
            print(f"It is of {z.genre} genre and {z.duration} minutes duration. ")
    @staticmethod
    def showAllBranchInfo():
        for k,v in StarCinema.all_branch_info.items():
            print(f"Branch name: {k}")
            s = 1
            for i in v:
                print(f'Movie No: {s}')
                s+=1
                print(i.movieInfo())
                print("****************************")
            print('#####################################')


movie1 = Movie('Matir Moyna', 'Liberation War Drama', 180)
movie2 = Movie('Inside Out', 'Animation', 114)
movie3 = Movie('The Secret Life of Walter Mitty', 'Adventure', 163)
print('1==========================================')
print(movie3.movieInfo())
print('2==========================================')
movie4 = Movie.createMovie_fromString('Shawshank Redemption-Thriller-153')
print('3==========================================')
print(movie4.movieInfo())
print('4==========================================')
branch1 = StarCinema('Mohakhali')
print('5==========================================')
branch1.addMovies(movie1, movie2, movie4)
print('6==========================================')
branch1.addMovies(movie1, movie3)
print('7==========================================')
StarCinema.showAllBranchInfo()
print('8==========================================')
branch2 = StarCinema('Mirpur')
print('9==========================================')
branch2.addMovies(movie1, movie2, movie3)
print('10==========================================')
StarCinema.showAllBranchInfo()
print('11==========================================')
StarCinema.check('Matir Moyna')
print('12=========================================')
StarCinema.check('Sound of Freedom')
print('13=========================================')
branch1.removeMovie(movie2)
StarCinema.showAllBranchInfo()
print('14=========================================')
branch2.removeMovie(movie1)
StarCinema.showAllBranchInfo()
```
